### Evaluation Findings

The FAISS retriever generally returned semantically related complaint chunks, especially for money transfer and credit card questions. However, some questions returned unrelated categories, showing that retrieval quality can vary depending on the available complaint data.

The FLAN-T5 Small generator produced very short answers in several cases. Although retrieved context was sometimes relevant, the model often failed to summarize multiple complaint details. Using a larger instruction-tuned model and improving prompt design could improve answer quality.
